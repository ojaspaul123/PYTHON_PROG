```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('fivethirtyeight')
```

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
#Scatter plot
sns.relplot(x = 'total_bill',y='tip',kind = 'scatter',data=df)
```

```python
#hue paramter
sns.relplot(x = 'total_bill',y='tip',hue = 'smoker',data=df)
```

```python
# Style Parameter
sns.relplot(x = 'total_bill',y='tip',hue = 'smoker', style = 'sex',data=df)
```

```python
# Size Parameter
sns.relplot(x = 'total_bill',y='tip',hue = 'smoker', style = 'sex',size ='size',sizes=(15,200),data=df)
```

```python
# SCATTER PLOT
sns.scatterplot(x = 'total_bill',y='tip',hue = 'smoker',data=df)
```

```python
fl = sns.load_dataset('flights')
```

```python
fl.head()
```

```python
x=fl.pivot_table(index='year',columns='month',values='passengers',aggfunc='sum')
```

```python
sns.heatmap(x)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True,fmt='d')
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True,fmt='d',cmap='winter')
```

```python
print("="*50)
print("CLUSTERMAPS")
print("="*50)
```

```python
sns.clustermap(x)
```

```python
sns.clustermap(x,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True,col_cluster=False,metric='correlation')
```
